```python

for col in ['tpep_pickup_datetime', 'tpep_dropoff_datetime']:
    df[col] = pd.to_datetime(df[col], errors='coerce')

# Create a new duration column
df['trip_duration'] = (df['tpep_dropoff_datetime'] - df['tpep_pickup_datetime']).dt.total_seconds() / 60


initial_rows = len(df)
valid_columns = [
    'tpep_pickup_datetime',
    'tpep_dropoff_datetime',
    'passenger_count',
    'trip_distance',
    'fare_amount',
    'total_amount',
    'trip_duration'
]
df = df.dropna(subset=valid_columns)

df = df[
    (df['trip_duration'] > 0) &
    (df['trip_distance'] > 0) &
    (df['passenger_count'] >= 1) &
    (df['fare_amount'] >= 0) &
    (df['total_amount'] >= 0)
]


duration_max = df['trip_duration'].quantile(0.99)
distance_max = df['trip_distance'].quantile(0.99)
fare_max = df['fare_amount'].quantile(0.99)
total_max = df['total_amount'].quantile(0.99)

df = df[
    (df['trip_duration'] <= duration_max) &
    (df['trip_distance'] <= distance_max) &
    (df['fare_amount'] <= fare_max) &
    (df['total_amount'] <= total_max)
]

cleaned_rows = len(df)
print(f"Rows before cleaning: {initial_rows}")
print(f"Rows after cleaning: {cleaned_rows}")
print(f"Dropped rows: {initial_rows - cleaned_rows}")
print("Current columns:", df.columns.tolist())
print("Trip duration range (min,max):", df['trip_duration'].min(), df['trip_duration'].max())


df.head()
```
